```python
def process_func(examples):
    tokenized_examples = tokenizer(
        text = examples["question"],
        text_pair=examples["context"],
        return_offsets_mapping=True,
        max_length=512,
        truncation="only_second",
        padding="max_length"
    )
    offset_mapping = tokenized_examples.pop("offset_mapping")
    start_positions = []
    end_positions = []
    for idx,offset in enumerate(offset_mapping):
        answer = examples["answers"][idx]
        # 答案的起始位置
        start_char = answer["answer_start"][0] 
        end_char = start_char + len(answer["text"][0]) 
        # 需要定位答案在token中的起始位置和结束位置
        # 一种策略，我们要拿到context的起始和结束，然后从左右两侧向答案逼近，要确保 start_char 和 end_char 都要在context中
        context_start = tokenized_examples.sequence_ids(idx).index(1)
        context_end = tokenized_examples.sequence_ids(idx).index(None,context_start)-1
        #获取了上下文的开始和结束位置
        # 判断答案是否在context中
        if offset[context_end][1]<start_char or offset[context_start][0]>end_char:
            start_token_pos = 0
            end_token_pos = 0
        else:
            # offset[token_id][0] 是第token_id个 token 的起始字符位置。
            # 循环从上下文的起始 token（context_start）开始，逐个向右移动，直到找到第一个起始位置大于或等于start_char的 token。
            token_id = context_start
            while token_id<=context_end and offset[token_id][0]<start_char:
                token_id+=1
            start_token_pos = token_id
            # 同理 offset[token_id][1] 是第token_id个 token 的终止字符位置。
            token_id = context_end
            while token_id>=context_start and offset[token_id][1]>end_char:
                token_id-=1
            end_token_pos = token_id
        start_positions.append(start_token_pos)
        end_positions.append(end_token_pos)
        
    tokenized_examples["start_positions"] = start_positions
    tokenized_examples["end_positions"] = end_positions
    return tokenized_examples
```

```python
datasets["train"].column_names
```

```python
tokenized_datasets = datasets.map(process_func, batched=True, remove_columns=datasets["train"].column_names)
tokenized_datasets
```

```python
#加载模型
model = AutoModelForQuestionAnswering.from_pretrained("hfl/chinese-macbert-base")
```

```python
args = TrainingArguments(
    output_dir="models_for_qa",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_steps=50,
    num_train_epochs=1
)
```

```python
trainer = Trainer(
    model=model,
    args=args,
    tokenizer=tokenizer,
    train_dataset=tokenized_datasets["train"].select(range(10)),
    eval_dataset=tokenized_datasets["validation"].select(range(10)),
    data_collator=DefaultDataCollator()
)
```

```python
trainer.train()
```
